# Attenuation at λ = 515, 532 and 550 nm,

## For λc = 502 nm

```python
# Reference attenuation parameters (assumed typical values)
alpha_0 = 0.2  # Reference attenuation in dB/m (estimated for slightly above cutoff)
lambda_s = 15   # Scaling factor in nm (empirical for visible fibers)

# Wavelengths of interest (in nm)
wavelengths = np.array([515, 532, 550])  
cutoff_wavelengths = np.array([502, 470, 350])

# Compute attenuation using the exponential model
def attenuation(lambda_c_nm, wavelengths):
    return alpha_0 * np.exp((wavelengths - lambda_c_nm) / lambda_s)    # in dB/m)  

attenuation_results = dict()
# Store results
for cutoff_wavelength in cutoff_wavelengths:
    attenuation_results[str(cutoff_wavelength) + ' nm'] = attenuation(cutoff_wavelength, wavelengths)

attenuation_results_df = pd.DataFrame(attenuation_results).T
attenuation_results_df.columns = wavelengths
attenuation_results_df.index.name = 'Cutoff Wavelength'
attenuation_results_df
```

## Expected output power

```python
# Input parameters
P_in = 6500  # Input power in µW
L = 5  # Fiber length in meters

# Function to calculate output power
def calculate_output_power(P_in, alpha, L):
    return P_in * 10 ** (-alpha * L / 10)

# Calculate expected output power for each case
output_powers = {}
for cutoff_wavelength, alphas in zip(cutoff_wavelengths, attenuation_results.values()):
    output_powers[cutoff_wavelength] = [calculate_output_power(P_in, alpha, L) for alpha in alphas]

output_powers_df = pd.DataFrame(output_powers).T
output_powers_df.columns = wavelengths
output_powers_df.index.name = 'Cutoff Wavelength'
output_powers_df
```
